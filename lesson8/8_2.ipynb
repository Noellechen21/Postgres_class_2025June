```python
#請建立一個範本list
my_list = [1,2,3,4,5]
print(type(my_list))
my_list.append(6)
print(my_list)
```

```python
#請建立一個範本list
my_list = [1, 2, 3, 4, 5]
#並且建立常用的實體方法
my_list.append(6)
print(my_list)
my_list.remove(2)
print(my_list)
my_list.insert(1, 10)
print(my_list)
my_list.sort()
print(my_list)
my_list.reverse()
print(my_list)
my_list.clear()
print(my_list)
```

```python

#我有一個list,要使用for迴圈,請將list中的每個元素乘以2
my_list = [1, 2, 3, 4, 5]
for i in my_list:
    print(i * 2)

for word in "Apple": #字串也是串列資料
    print(word)
```

```python
#請建立國家code與國家名稱的字典
country_dict = {
    "TW": "Taiwan",
    "US": "United States",
    "JP": "Japan",
    "CN": "China",
    "KR": "South Korea"
}
```

```python
 country_dict.keys()  #用實體名稱.實體方法執行 發現結果像一個list,而list是串列資料>for迴圈
```

```python
###
for code in country_dict.keys():
    print(code, country_dict[code])
```

```python
for country in country_dict.values():
    print(country)
```

```python
for item in country_dict.items():
    print(item)

    #執行結果是tuple只是暫時的，整個可以拿來當區塊變數
```

```python
for key, value in country_dict.items():
    print(key, value)
```
