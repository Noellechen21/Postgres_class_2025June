```python
 #將帶有密碼的環境變數隱藏(從9_3複製過來的指令，直接改)
import psycopg2
from dotenv import load_dotenv  #0810修改_dotnev建議的操作法
load_dotenv()  #0810 模擬環境變數
import os #0810修改

#請幫我建立一個function
#傳入connection參數
#建立一個cursor
#執行SQL查詢
#並回傳查詢結果

def execute_query(connection, query):
    cursor = connection.cursor()
    cursor.execute(query)
    result = cursor.fetchall()
    cursor.close()
    return result
# 0810 使用python內建函式抓環境變數
def create_connection():
    conn = psycopg2.connect(
        host=os.getenv("HOST"),  #這稱為引數名稱的呼叫
        database=os.getenv("DATABASE"),
        user=os.getenv("USER"),
        password=os.getenv("PASSWORD"),
        port="5432"
    )
    return conn


def main():
    conn = create_connection()
    print(type(conn)) #檢查conn的功能
    if conn:
        print("成功連接到資料庫！")
        query = """
        SELECT count(*) AS "筆數"
        FROM "台鐵車站資訊";
        """
        result = execute_query(conn, query)
        print("台鐵車站資訊：", result)
        conn.close()
    else:
        print("無法連接到資料庫，請檢查設定。")
        return

if __name__ == "__main__":
    main()
```
